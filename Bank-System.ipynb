```python
import random
```

```python
all_accounts = []
```

```python
def open_account():
    account_title = input("Account Title:")
    cnic = input("CNIC:")
    contact_no = input("Contact Number:")
    initial_deposit = int(input("Initial Amount:"))
    account_no = random.randint(11111111,99999999)
    all_accounts.append({"Account_No":account_no,"Account_Title":account_title,"Contact":contact_no,
    "CNIC":cnic,"Balance":initial_deposit})
    print("Your Account Opened!")
    print(f"Your account title is {account_title} and account number is {account_no}.")
```

```python
def account_details(account_no):
    for account in all_accounts:
        if account["Account_No"]==account_no:
            print(f"""
            ================================
                       ABC Bank
            ================================
            Account_Number : {account["Account_No"]}
            Title : {account["Account_Title"]}
            CNIC : {account["CNIC"]}
            Contact_Number : {account["Contact"]}
            Balance : {account["Balance"]}
            
            """)
    # else:
    #     print("Account Not Found!")
```

```python
def cash_deposit(account_no,amount):
    for account in all_accounts:
        if account["Account_No"]==account_no:
            if amount>0:
                account["Balance"]+=amount
                print("Deposit Successfully")
                break
            else:
                print("Insufficient Amount")
                break
    else:
        print("Account Not Found!")
```

```python
def cash_withdraw(account_no,amount):
    for account in all_accounts:
        if account["Account_No"]==account_no:
            if amount<account["Balance"]:
                account["Balance"]-=amount
                print("Withdraw Successfully")
                break
            else:
                print("Insufficient Amount")
                break
    else:
        print("Account Not Found!")
```

```python
def check_amount(account_no):
    for account in all_accounts:
        if account["Account_No"]==account_no:
            return f"Your account balance is {account["Balance"]}"
    else:
        print("Invalid Account Number!")
```

```python
def close_account(account_no):
    for index,account in enumerate(all_accounts):
        if account["Account_No"]==account_no:
            del all_accounts[index]
            print("Account Removed!")
            break
    else:
        print("Account Not Found!")
```

```python
def amount_transfer(sender_acc,amount,reciever_acc):
    for sender_account in all_accounts:
        if sender_account["Account_No"]==sender_acc:
            for receiver_account in all_accounts:
                if receiver_account["Account_No"]==reciever_acc:
                    receiver_account["Balance"]+=amount
                    sender_account["Balance"]-=amount
                    print("Amount Transfer Successfully.")
                    break 
            else:
                print("Reciever Account Not Found!")
            break
    else:
        print("Sender Account Not Found!")
```

```python
while True:
    print(f"""
    ================================
                MCB Bank
    ================================

    -------------------------
    1. Open Account
    2. Show Account Details
    3. Cash Deposit
    4. Cash Withdraw
    5. Check Amount
    6. Send Money
    7. Close Account
    8. Break
    -------------------------""")
    choice = int(input("Enter the option:"))
    if choice ==1:
        open_account()
    elif choice==2:
        account_no = int(input("Enter your account number show all detail."))
        account_details(account_no)
    elif choice==3:
        account_no = int(input("Enter your account number."))
        amount = int(input("Enter the amount which youu want to deposit"))
        cash_deposit(account_no,amount)
    elif choice==4:
        account_no = int(input("Enter your account number."))
        amount = int(input("Enter the amount which youu want to deposit"))
        cash_withdraw(account_no,amount)
    elif choice==5:
        account_no = int(input("Enter your account number."))
        message = check_amount(account_no)
        print(message)
    elif choice==6:
        sender_acc = int(input("Sender Account Number:"))
        amount = int(input("Enter the amount which youu want to transfer"))
        reciever_acc = int(input("Receiver Account Number:"))
        amount_transfer(sender_acc,amount,reciever_acc)
    elif choice==7:
        account_no = int(input("Enter your account number."))
        close_account(account_no)
    elif choice==8:
        print("Thanks for using MCB mobile banking service.")
        break
```

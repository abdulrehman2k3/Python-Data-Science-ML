```python
import pandas as pd
import numpy as np
import warnings

warnings.filterwarnings("ignore")
```

```python
data = pd.read_csv("insurance.csv")
```

```python
data
```

```python
data.sex=data.sex.replace({"male":1,"female":0})
```

```python
data
```

```python
data.smoker.replace({"yes":1,"no":0})
```

```python
data.region.unique()
```

```python
region_dumies = pd.get_dummies(data.region).astype(int)
```

```python
data = pd.concat([data,region_dumies],axis=1)
```

```python
data.drop(["region"],axis=1,inplace=True)
```

```python
data
```

# Multiple Regression

```python
x = data.drop(["charges"],axis=1)
y = data.charges
```

```python
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.20,random_state=42)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
from sklearn.linear_model import LinearRegression
```

```python
mlr = LinearRegression()
```

```python
mlr.fit(x_train,y_train)
```

```python
mlr.coef_
```

```python
mlr.intercept_
```

```python
y_pred_train = mlr.predict(x_train)
```

```python
y_pred_train
```

```python
from sklearn.metrics import r2_score,root_mean_squared_error
```

```python
root_mean_squared_error(y_train,y_pred_train)
```

```python
r2_score(y_train,y_pred_train)
```

```python
y_pred_test = mlr.predict(x_test)
```

```python
root_mean_squared_error(y_test,y_pred_test)
```

```python
r2_score(y_test,y_pred_test)
```

```python
def predict_charges(data): 
    return mlr.predict(data)
```

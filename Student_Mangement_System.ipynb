# Student Management System:
  - Add Students
  - View Students
  - Update Student 
  - Delete Student
  - Search Student
  - View All Records

```python
import pandas as pd 
class student_management_system():
    def add_students(self):
        name = input("Enter your full name: ")
        roll_no = input("Enter your roll no: ")
        age = int(input("Enter your age: "))
        grade = int(input("Enter your class: "))
        total_marks = int(input("Enter your total_marks here: "))
        all_students.append({"Roll_No":roll_no,
                             "Name":name,
                             "Age":age,
                             "Class":grade,
                             "Total_Marks":total_marks
                            })
        return f"{name} added successfully."
    def update_student(self):
        roll_no = input("Enter your roll no: ")
        for student in all_students:
            if roll_no == student["Roll_No"]:
                name = input("Enter your updated name: ")
                age = int(input("Enter your updated age: "))
                grade = int(input("Enter your updated class: "))
                total_marks = int(input("Enter your updated marks: "))
                student["Name"]=name
                student["Age"]=age
                student["Class"]=grade
                student["Total_Marks"]=total_marks
                return "Student detail update successfully"
        else:
            return "Not Exist!"
    def delete_student(self):
        roll_no = input("Enter your roll no: ")
        found = False
        for student in all_students:
            if roll_no == student["Roll_No"]:
                all_students.remove(student)
                return "Student Remove From System!"
                found = True
                break
        if not found:
            print("Student Not Found!")
    def search_student(self):
        full_name = input("Enter your full name: ")
        student_record=[]
        for student in all_students:
            if full_name in student["Name"]:
                student_record.append(student)    
                return pd.DataFrame(student_record)
        else:
            return "Not Exist!"
    def view_all_records(self):
        records = []
        for student in all_students:
            records.append(student)
        return pd.DataFrame(records)
student = student_management_system()
while True:
    print("===================Student Management System=========================")
    print("1. Add Students")
    print("2. Update Student")
    print("3. Delete Student Record")
    print("4. Search Student")
    print("5. View All Records")
    print("6. Exit")
    choice = int(input("Enter the option(1-6):"))
    if choice==1:
        print(student.add_students())
    elif choice==2:
        print(student.update_student())
    elif choice==3:
        print(student.delete_student())
    elif choice == 4:
        print(student.search_student())
    elif choice==5:
        print(student.view_all_records())
    elif choice == 6:
        print("Thanks for using it.")
        break
```

```python
all_tasks = []
def add_tasks():
    print("\n📝 Add your tasks below (type 'stop' when done)")
    while True:
        task = input("➡️  Enter your today's task: ").lower()
        if task=="stop":
            break
        all_tasks.append(task)
    print("✅ Tasks added successfully!\n")

def completed_task():
    completed_tasks = {}
    if not all_tasks:
        print("⚠️ No tasks to check! Add some tasks first.\n")
        return
    for task in all_tasks:
        answer = input(f"Is '{task}' completed? (Yes/No): ").lower()
        if answer=="yes":
            completed_tasks[task] = "✅ Completed"
        elif answer=="no":
            completed_tasks[task] = "❌ Incompleted"
        else:
            print("⚠️ Invalid input, marking as Incompleted")
            completed_tasks[task] = "❌ Incompleted"
    print(completed_tasks)

def delete_task():
    if not all_tasks:
        print("⚠️ No task to remove! List is already empty.\n")
        return
    del_task = input("🗑️  Enter task which you want to delete: ").lower()
    for task in all_tasks:
        if task == del_task:
            all_tasks.remove(task)
            print(f"✅ '{task}' removed successfully!\n")
            return
    else:
        print(f"❌ '{del_task}' not found in your list.\n")

def show_tasks():
    if not all_tasks:
        print("Your list is empty! Add some tasks first.\n")
        return
    print("All Tasks:",all_tasks)

while True:
    print("============TO DO LIST============")
    print("1. ➕ Add Tasks")
    print("2. 🗑️ Delete Task")
    print("3. 📋 Show Tasks")
    print("4. ✅ Complete Task")
    print("5. 🚪 Exit")
    choice = int(input("👉 Enter your choice (1-5): "))
    if choice == 1:
        add_tasks()
    elif choice==2:
        delete_task()
    elif choice==3:
        show_tasks()
    elif choice==4:
        completed_task()
    elif choice==5:
        print("👋 Thanks for using To-Do List Manager. Goodbye!")
        break
    else:
        print("⚠️ Invalid choice! Please enter 1-5.\n")
```
